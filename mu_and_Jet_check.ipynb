```python
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt
from coffea.nanoevents import NanoEventsFactory as NF, NanoAODSchema
```

```python
import uproot

f = uproot.open(r"Z:\Project files\3f1aaaee-242f-44e4-a3d6-56bc2f6fb19b.root")

f.keys()
#done this for checking the file 
```

```python
events= NF.from_root(r"Z:\Project files\3f1aaaee-242f-44e4-a3d6-56bc2f6fb19b.root:Events",
                         schemaclass=NanoAODSchema).events()
```

```python
muon= events.Muon
```

```python
plt.figure(figsize=(18, 5))
plt.subplot(1,3,1)

plt.hist(ak.flatten(muon.pt),bins=50,range=(0,200) , color='b',alpha=0.5)
plt.title("$\mu$ pT")
plt.xlabel("pT (GeV)")
plt.ylabel("Events")

plt.subplot(1,3,2)
plt.hist(ak.flatten(muon.eta),bins=50,range=(-4,4) , color='g',alpha=0.5)
plt.title("$\mu$ $\eta$")
plt.xlabel("eta")
plt.ylabel("Events")

plt.subplot(1,3,3)
plt.hist(ak.flatten(muon.phi),bins=50,range=(-5,5) , color='r',alpha=0.5)
plt.title("$\mu$ $\phi$")
plt.xlabel("phi")
plt.ylabel("Events")

plt.tight_layout()
plt.savefig("Jetcheck.png", dpi=300, bbox_inches='tight')
plt.show()
```

```python
jet= events.Jet
```

```python
plt.figure(figsize=(18, 5))
plt.subplot(1,3,1)

plt.hist(ak.flatten(jet.pt),bins=50,range=(0,200) , color='y',alpha=0.5)

plt.title("Jet pT")
plt.xlabel("pT (GeV)")
plt.ylabel("Events")


plt.subplot(1,3,2)
plt.hist(ak.flatten(jet.eta),bins=50,range=(-4,4) , color='purple',alpha=0.5)
plt.title("Jet $\eta$")
plt.xlabel("eta")
plt.ylabel("Events")


plt.subplot(1,3,3)
plt.hist(ak.flatten(jet.phi),bins=50,range=(-5,5) , color='gray',alpha=0.5)
plt.title("Jet $\phi$")
plt.xlabel("phi")
plt.ylabel("Events")


plt.tight_layout()
plt.savefig("Jetcheck.png", dpi=300, bbox_inches='tight')
plt.show()
```
